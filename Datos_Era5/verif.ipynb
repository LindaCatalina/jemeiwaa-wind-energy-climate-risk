```python
#python - <<'PY'
import xarray as xr
ds = xr.open_dataset("era5_guajira_daily_1981_2014.nc")
print(ds)
print("\nRango de fechas:", str(ds.time.values[0]), "->", str(ds.time.values[-1]))
print("Días totales:", ds.dims["time"])
print("\nVariables:", list(ds.data_vars))
# Estadísticos rápidos para ver números razonables
print("\nwind10 media/std:", float(ds.wind10.mean()), float(ds.wind10.std()))
print("t2m media (K):", float(ds.t2m.mean()))
print("sp media (Pa):", float(ds.sp.mean()))
#PY
```

```python
import xarray as xr
import numpy as np
ds = xr.open_dataset("era5_guajira_daily_1981_2014.nc")
dt = np.diff(ds.time.values).astype('timedelta64[D]').astype(int)
print("Saltos únicos en días:", set(dt.tolist()))
```

```python
import xarray as xr
ds = xr.open_dataset("era5_guajira_daily_1981_2014.nc")
print("Inicio/fin:", ds.time.isel(time=0).values, ds.time.isel(time=-1).values)
print("N días:", ds.time.size)
```

```python
import xarray as xr
ds = xr.open_dataset("era5_guajira_daily_1981_2014_150m.nc")
print(ds)
print("Inicio/fin:", ds.time.isel(time=0).values, ds.time.isel(time=-1).values)
print("N días:", ds.time.size)
print("Vars:", list(ds.data_vars))
```

```python
import xarray as xr, numpy as np
ds = xr.open_dataset("era5_guajira_daily_1981_2014_150m.nc")
dt = np.diff(ds.time.values).astype('timedelta64[D]').astype(int)
print("Saltos únicos en días:", set(dt.tolist()))
```

```python
import xarray as xr
ds = xr.open_dataset("era5_guajira_daily_1981_2014_150m.nc")
print("wind10 mean/std:", float(ds.wind10.mean()), float(ds.wind10.std()))
print("v150_log mean/std:", float(ds.v150_log.mean()), float(ds.v150_log.std()))
print("rho mean/std:", float(ds.rho.mean()), float(ds.rho.std()))
```

```python
import xarray as xr
ds = xr.open_dataset("era5_guajira_daily_1981_2014_150m.nc")
ratio = (ds.v150_log / ds.wind10).median().item()
print("Mediana v150_log / wind10:", ratio)
```
