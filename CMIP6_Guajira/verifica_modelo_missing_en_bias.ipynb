```python
import xarray as xr
from pathlib import Path
import numpy as np
import re

root = Path(".")
bias_dir = root / "bias_factors"

# Modelos con CSV de factores
csv_models = {p.name.replace("_bias_factors.csv", "") for p in bias_dir.glob("*_bias_factors.csv")}

# Modelos con viento (sfcWind) o con pares uas+vas (histórico)
sfc_models = {re.match(r"([^_]+)_", p.name).group(1) for p in root.glob("*_historical_sfcWind_*_guajira.nc")}
ua_models = {re.match(r"([^_]+)_", p.name).group(1) for p in root.glob("*_historical_uas_*_guajira.nc")}
va_models = {re.match(r"([^_]+)_", p.name).group(1) for p in root.glob("*_historical_vas_*_guajira.nc")}
wind_models = sfc_models | (ua_models & va_models)

missing = sorted(wind_models - csv_models)
print("Modelos con viento que NO tienen bias_factors:", missing)

def check_var(model, exp, var):
    files = list(root.glob(f"{model}_{exp}_{var}_*_guajira.nc"))
    if not files:
        return False, "sin archivo"
    try:
        ds = xr.open_dataset(files[0], use_cftime=True)
        t = ds["time"]
        yrs = (t.dt.year.min().item(), t.dt.year.max().item())
        return True, f"ok ({yrs[0]}–{yrs[1]})"
    except Exception as e:
        return False, f"error: {e}"

for m in missing:
    print(f"\n=== {m} ===")
    # solape histórico
    files_hist = list(root.glob(f"{m}_historical_*_guajira.nc"))
    hist_years = "n/a"
    for f in files_hist:
        try:
            t = xr.open_dataset(f, use_cftime=True)["time"]
            hist_years = f"{t.dt.year.min().item()}–{t.dt.year.max().item()}"
            break
        except Exception:
            continue
    print("histórico (cualquier var) años:", hist_years)
    for exp in ["historical", "ssp126", "ssp245", "ssp585"]:
        ok_tas, msg_tas = check_var(m, exp, "tas")
        ok_ps, msg_ps = check_var(m, exp, "ps")
        print(f" {exp}: tas {msg_tas} | ps {msg_ps}")
```
